#**DATA MINING**


##CARREGANDO OS DATASETS

```python
import pandas as pd
```

```python
import glob
import pandas as pd

# ==========================
# CARREGAR INFRAÇÕES
# ==========================

arquivos_infra = glob.glob('/content/datasets/infracoes/*.csv')

dfs_infra = [
    pd.read_csv(
        arq,
        sep=';',
        encoding='latin1'
    )
    for arq in arquivos_infra
]

infra = pd.concat(
    dfs_infra,
    ignore_index=True
)

# ==========================
# CARREGAR ACIDENTES
# ==========================

arquivos_acidentes = glob.glob('/content/datasets/acidentes/*.csv')

dfs_acidentes = [
    pd.read_csv(
        arq,
        sep=';',
        encoding='latin1'
    )
    for arq in arquivos_acidentes
]

acidentes = pd.concat(
    dfs_acidentes,
    ignore_index=True
)

# ==========================
# VERIFICAÇÃO
# ==========================

print("INFRAÇÕES")
print(infra.shape)

print("\nACIDENTES")
print(acidentes.shape)

print("\nPrimeiras linhas de infrações")
display(infra.head())

print("\nPrimeiras linhas de acidentes")
display(acidentes.head())
```

##VISUALIZANDO COLUNAS E TIPOS

```python
print("COLUNAS INFRAÇÕES")
print(infra.columns.tolist())

print("\nCOLUNAS ACIDENTES")
print(acidentes.columns.tolist())
```

```python
infra.info()

print("\n" + "="*50 + "\n")

acidentes.info()
```

##UNIFICANDO DATASETS DE INFRAÇÕES

```python
dfs_infra = []

for arquivo in arquivos_infra:

    df = pd.read_csv(
        arquivo,
        sep=';',
        encoding='latin1'
    )

    dfs_infra.append(df)
```

```python
print(len(dfs_infra))
```

```python
infra = pd.concat(
    dfs_infra,
    ignore_index=True
)
```

```python
print(infra.columns.tolist())
```

###SALVA DATASET UNIFICADO

```python
infra.to_csv(
    '/content/infra_unificado.csv',
    index=False
)
```
